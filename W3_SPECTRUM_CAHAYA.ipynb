1. perhitungan variasi 30 cm lampu pijar

```python
import numpy as np
import pandas as pd

# Data dari tabel lampu pijar variasi 30 cm
warna = ["ungu", "biru", "cyan", "hijau", "kuning", "oranye", "merah"]
jarak_l = 30  # cm (l pada rumus)
d = 0.002  # cm
l_kanan = np.array([6.2, 7, 8, 8, 9, 9.5, 10.5])  # cm
l_kiri = np.array([6.2, 7, 8, 8, 8.8, 9.6, 10.5])  # cm

# Menghitung lambda untuk L kanan dan L kiri
lambda_kanan = (d * l_kanan) / np.sqrt(jarak_l**2 + l_kanan**2)
lambda_kiri = (d * l_kiri) / np.sqrt(jarak_l**2 + l_kiri**2)


# Menghitung rata-rata lambda
lambda_rata_rata = (lambda_kanan + lambda_kiri) / 2

# Konversi dari cm ke nm (1 cm = 10^6 nm)
lambda_kanan_nm = lambda_kanan * 1e6
lambda_kiri_nm = lambda_kiri * 1e6
lambda_rata_rata_nm = lambda_rata_rata * 1e6


# Membuat DataFrame untuk hasil
df = pd.DataFrame({
    "Warna": warna,
    "Jarak L (cm)": [jarak_l] * len(warna),
    "L Kanan (cm)": l_kanan,
    "L Kiri (cm)": l_kiri,
    "Lambda Kanan (nm)": lambda_kanan_nm,
    "Lambda Kiri (nm)": lambda_kiri_nm,
    "Lambda Rata-rata (nm)": lambda_rata_rata_nm})


# Menampilkan hasil
print(df)
```

```python
import numpy as np
import pandas as pd

# Data eksperimen pada jarak 30 cm
warna = ["ungu", "biru", "cyan", "hijau", "kuning", "oranye", "merah"]
l = 30  # cm (jarak antara celah dan layar)
d = 0.002  # cm (jarak antar celah)
x_kanan = np.array([6.2, 7, 8, 8, 9, 9.5, 10.5])  # cm
x_kiri = np.array([6.2, 7, 8, 8, 8.8, 9.6, 10.5])  # cm

# Nilai panjang gelombang teori (dalam nm)
lambda_teori = np.array([402.8, 447.3,483.1, 537.9,568.6,606.4,639.8])  # dalam nm

# Menghitung lambda eksperimen (perhitungan)
lambda_kanan = (d * x_kanan) / np.sqrt(l**2 + x_kanan**2)
lambda_kiri = (d * x_kiri) / np.sqrt(l**2 + x_kiri**2)
lambda_rata_rata = (lambda_kanan + lambda_kiri) / 2

# Konversi dari cm ke nm (1 cm = 10^6 nm)
lambda_rata_rata_nm = lambda_rata_rata * 1e6

# Menghitung ralat mutlak (dalam %)
ralat_mutlak = np.abs((lambda_teori - lambda_rata_rata_nm) / lambda_rata_rata_nm) * 100

# Membuat DataFrame untuk hasil
df = pd.DataFrame({
    "Warna": warna,
    "Lambda Teori (nm)": lambda_teori,
    "Lambda Eksperimen (nm)": lambda_rata_rata_nm,
    "Ralat Mutlak (%)": ralat_mutlak
})

# Menampilkan hasil
print(df)
```
