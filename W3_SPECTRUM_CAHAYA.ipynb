```python
import numpy as np
import pandas as pd

warna = ["ungu", "biru", "cyan", "hijau", "kuning", "oranye", "merah"]

l = 30 #cm
d = 0.002

x_kanan = np.array([6.6, 7.4, 7.6, 8, 8.8, 9, 9.7]) #cm
x_kiri  = np.array([6.6, 7.4, 7.4, 8, 8.9, 9, 10.3])

lambda_teori = np.array([402.8, 447.3, 483.1, 537.9, 568.6, 606.4, 639.8])

lambda_kanan = (d * x_kanan) / np.sqrt(l**2 + x_kanan**2)
lambda_kiri  = (d * x_kiri)  / np.sqrt(l**2 + x_kiri**2)
lambda_rata_rata = (lambda_kanan + lambda_kiri) / 2

lambda_kanan_nm = lambda_kanan * 1e6
lambda_kiri_nm  = lambda_kiri * 1e6
lambda_rata_nm  = lambda_rata_rata * 1e6

ralat_mutlak = np.abs((lambda_rata_nm-lambda_teori) / lambda_teori) * 100

df = pd.DataFrame({
    "Warna": warna,
    "L (cm)": [l]*len(warna),
    "x kanan (cm)": x_kanan,
    "x kiri (cm)": x_kiri,
    "Lambda kanan (nm)": lambda_kanan_nm,
    "Lambda kiri (nm)": lambda_kiri_nm,
    "Lambda rata-rata (nm)": lambda_rata_nm,
    "Lambda teori (nm)": lambda_teori,
    "Ralat (%)": ralat_mutlak
})

print(df)
```
