```python
import matplotlib.pyplot as plt
import numpy as np

sudut = np.array([20, 30, 40, 50, 52, 54, 54.2, 54.4, 54.6, 54.8, 55, 55.2, 55.4, 55.6, 55.8, 56, 58, 60, 70, 80])
sejajar = np.array([31.0, 43.0, 14.9, 5.0, 1.8, 0.2, 0.1, 0.2, 0.2, 0.2, 0.2, 0.4, 0.5, 0.7, 0.5, 0.3, 1.6, 6.3, 8.4, 0.4])
tegak_lurus = np.array([24.7, 29.6, 35.1, 99.6, 54.6, 58.1, 58.3, 81.0, 60.6, 60.7, 60.0, 64.1, 64.8, 71.2, 71.3, 68.8, 91.5, 83.5, 29.4, 53.5])

idx_min = np.argmin(sejajar)
theta_b = sudut[idx_min]
tegangan_min = sejajar[idx_min]

n1 = 1.0  #Indeks bias udara
n2 = np.tan(np.radians(theta_b)) * n1

print(f"--- HASIL ANALISIS DATA ---")
print(f"Tegangan Minimum (Sejajar) : {tegangan_min} mV")
print(f"Sudut Brewster (theta_b)   : {theta_b}°")
print(f"Indeks Bias Kaca (n2)      : {n2:.4f}")

plt.figure(figsize=(12, 7))
plt.plot(sudut, sejajar, 'b-o', label='Polarisasi Sejajar (p)', linewidth=2)
plt.plot(sudut, tegak_lurus, 'r-o', label='Polarisasi Tegak Lurus (s)', linewidth=2, alpha=0.6)

plt.axvline(x=theta_b, color='green', linestyle='--', label=f'Sudut Brewster ({theta_b}°)')
plt.scatter(theta_b, tegangan_min, color='black', s=70, zorder=5)

plt.title('Grafik Analisis Perbandingan Tegangan terhadap Sudut Refleksi', fontsize=14)
plt.xlabel('Sudut Refleksi (°)', fontsize=12)
plt.ylabel('Tegangan (mV)', fontsize=12)
plt.legend()
plt.grid(True, which='both', linestyle=':', alpha=0.5)
plt.show()
```
