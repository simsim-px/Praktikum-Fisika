```python
import matplotlib.pyplot as plt
import numpy as np

sudut_rep1 = [20, 30, 40, 50, 52, 54, 54.2, 54.4, 54.6, 54.8, 55, 55.2, 55.4, 55.6, 55.8, 56, 58, 60, 70, 80]
voltages_rep1_mV = [31.0, 43.0, 14.9, 5.0, 1.8, 0.2, 0.1, 0.2, 0.2, 0.2, 0.2, 0.4, 0.5, 0.7, 0.5, 0.3, 1.6, 6.3, 84.8, 452]

sudut_rep2 = [20, 30, 40, 50, 52, 54, 54.2, 54.4, 54.6, 54.8, 55, 55.2, 55.4, 55.6, 55.8, 56, 58, 60, 70, 80]
voltages_rep2_mV = [24.7, 29.6, 35.1, 99.6, 54.6, 58.1, 58.3, 81.0, 60.6, 60.7, 60.0, 64.1, 64.8, 71.2, 71.3, 68.8, 91.5, 83.5, 294, 535]

voltages_rep1 = [v / 1000 for v in voltages_rep1_mV]
voltages_rep2 = [v / 1000 for v in voltages_rep2_mV]

start_angle = 20
end_angle = 80

indices_rep1 = [i for i, angle in enumerate(sudut_rep1) if start_angle <= angle <= end_angle]
voltages_subset_rep1 = [voltages_rep1[i] for i in indices_rep1]
min_voltage_original_index_rep1 = indices_rep1[np.argmin(voltages_subset_rep1)]

indices_rep2 = [i for i, angle in enumerate(sudut_rep2) if start_angle <= angle <= end_angle]
voltages_subset_rep2 = [voltages_rep2[i] for i in indices_rep2]
min_voltage_original_index_rep2 = indices_rep2[np.argmin(voltages_subset_rep2)]

plt.figure(figsize=(10, 6))
plt.plot(sudut_rep1, voltages_rep1, marker='o', linestyle='-', label='Sejajar')
plt.plot(sudut_rep2, voltages_rep2, color='purple', marker='o', linestyle='--', label='Tegak lurus')

plt.plot(sudut_rep1[min_voltage_original_index_rep1], voltages_rep1[min_voltage_original_index_rep1],
         marker='.', markersize=15, color='red', label=f"Sudut Brewster: {sudut_rep1[min_voltage_original_index_rep1]}°")

plt.title('Grafik Tegangan terhadap Sudut dalam Kondisi Polarisasi Sejajar')
plt.xlabel('Sudut (°)')
plt.ylabel('Tegangan (V)')
plt.grid(True)
plt.legend()
plt.show()
```
