```python
import matplotlib.pyplot as plt
import numpy as np

# Data for repetition 1
angles_rep1 = [20, 30, 40, 50, 52, 54, 56, 58, 60, 70]
voltages_rep1 = [0.5, 0.8, 1.4, 1.8, 1.7, 0.8, 2.1, 2.3, 2.1, 4.0]

# Data for repetition 3
angles_rep2 = [20, 30, 40, 50, 52, 54, 56, 58, 60, 70]
voltages_rep2 = [0.9, 1.0, 1.1, 1.5, 1.4, 0.9, 1.9, 2.1, 1.9, 4.0]

# Define angle range
start_angle = 20
end_angle = 60

# Find and mark minimum voltage for rep 1
indices_rep1 = [i for i, angle in enumerate(angles_rep1) if start_angle <= angle <= end_angle]
voltages_subset_rep1 = [voltages_rep1[i] for i in indices_rep1]
min_voltage_index_in_subset_rep1 = np.argmin(voltages_subset_rep1)
min_voltage_original_index_rep1 = indices_rep1[min_voltage_index_in_subset_rep1]

# Find and mark minimum voltage for rep 2
indices_rep2 = [i for i, angle in enumerate(angles_rep2) if start_angle <= angle <= end_angle]
voltages_subset_rep2 = [voltages_rep2[i] for i in indices_rep2]
min_voltage_index_in_subset_rep2 = np.argmin(voltages_subset_rep2)
min_voltage_original_index_rep2 = indices_rep2[min_voltage_index_in_subset_rep2]

# Plotting
plt.figure(figsize=(10, 6))

plt.plot(angles_rep1, voltages_rep1, marker='o', linestyle='-', label='Pengulangan 1')
plt.plot(angles_rep2, voltages_rep2, marker='x', linestyle='--', label='Pengulangan 2')

# Mark the minimum points
plt.plot(angles_rep1[min_voltage_original_index_rep1], voltages_rep1[min_voltage_original_index_rep1],
         marker='.', markersize=15, color='red', label='Min Voltage Rep 1')
plt.plot(angles_rep2[min_voltage_original_index_rep2], voltages_rep2[min_voltage_original_index_rep2],
         marker='.', markersize=15, color='purple', label='Min Voltage Rep 2')

plt.title('Grafik Tegangan terhadap Sudut (Data Praktikum Brewster Sejajar)')
plt.xlabel('Sudut (°)')
plt.ylabel('Tegangan (mV)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np

# Data sesuai gambar
angles = [20, 30, 40, 50, 52, 54, 56, 58, 60, 70]
voltages_rep1 = [0.5, 0.8, 1.4, 1.8, 1.7, 0.8, 2.1, 2.3, 2.1, 4.0]
voltages_rep2 = [0.9, 1.0, 1.1, 1.5, 1.4, 0.9, 1.9, 2.1, 1.9, 4.0]

# Buat DataFrame
data = {
    'Sudut': angles,
    'Pengulangan 1': voltages_rep1,
    'Pengulangan 2': voltages_rep2
}
df = pd.DataFrame(data)

# Hitung rata-rata dan standar deviasi
df['Mean Voltage'] = df[['Pengulangan 1', 'Pengulangan 2']].mean(axis=1)
df['Standard Deviation'] = df[['Pengulangan 1', 'Pengulangan 2']].std(axis=1)

# Tampilkan DataFrame
print(df)
```
